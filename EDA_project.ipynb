```python
#Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("ggplot")
```

```python
#Load Dataset
url = "https://raw.githubusercontent.com/salemprakash/EDA/main/Data/covid-data.csv"

covid = pd.read_csv(url)

covid.head()
```

```python
#Dataset Information
print("Shape:", covid.shape)

covid.info()

covid.describe(include="all")
```

```python
#Basic Statistical Analysis
covid.describe()

print("\nMean")
print(covid.mean(numeric_only=True))

print("\nMedian")
print(covid.median(numeric_only=True))

print("\nStandard Deviation")
print(covid.std(numeric_only=True))
```

```python
#Missing Values
print(covid.isnull().sum())

print("\nPercentage Missing")

print((covid.isnull().sum()/len(covid))*100)
```

```python
#Handle Missing Data
numeric_cols = covid.select_dtypes(include=np.number).columns

covid[numeric_cols] = covid[numeric_cols].fillna(
    covid[numeric_cols].median()
)

categorical_cols = covid.select_dtypes(include="object").columns

covid[categorical_cols] = covid[categorical_cols].fillna("Unknown")

covid.isnull().sum()
```

```python
#Data Cleaning
covid.drop_duplicates(inplace=True)

covid["date"] = pd.to_datetime(covid["date"])

covid = covid[
    (covid["total_cases"] >= 0) &
    (covid["total_deaths"] >= 0) &
    (covid["new_cases"] >= 0)
]

covid.head()
```

```python
#Data Transformation
covid["log_total_cases"] = np.log1p(covid["total_cases"])

covid["mortality_rate"] = (
    covid["total_deaths"] /
    covid["total_cases"]
) * 100

covid["mortality_rate"] = covid["mortality_rate"].fillna(0)

covid.head()
```

```python
# Convert date to datetime
covid["date"] = pd.to_datetime(
    covid["date"],
    errors="coerce"
)

# Sort by country and date
covid_sorted = covid.sort_values(
    ["location", "date"]
)

# Get the latest observation for each country
country_data = (
    covid_sorted
    .groupby("location", as_index=False)
    .last()
)

print("Number of countries:", len(country_data))
print("Dataset shape:", country_data.shape)

print("\nCountries:")
print(country_data["location"].tolist())
country_data[["location", "date"]].head(20)
```

```python
cluster_features = [
    "total_cases_per_million",
    "total_deaths_per_million",
    "people_vaccinated_per_hundred",
    "people_fully_vaccinated_per_hundred",
    "population_density",
    "median_age",
    "gdp_per_capita",
    "life_expectancy",
    "human_development_index"
]

# Keep only features that exist in the provided dataset
cluster_features = [
    col for col in cluster_features
    if col in country_data.columns
]

print("Features used for clustering:")
print(cluster_features)
```

```python
cluster_data = country_data[
    ["location"] + cluster_features
].copy()

print("Cluster dataset shape:", cluster_data.shape)

cluster_data
```

```python
print("Missing values:")
print(cluster_data[cluster_features].isnull().sum())
```

```python
#imputation for missing values

for col in cluster_features:
    cluster_data[col] = cluster_data[col].fillna(
        cluster_data[col].median()
    )

print("Total missing values remaining:",
      cluster_data[cluster_features].isnull().sum().sum())
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(
    cluster_data[cluster_features]
)

print("Number of samples:", X_scaled.shape[0])
print("Number of features:", X_scaled.shape[1])
```

```python
#elbow method for k means

from sklearn.cluster import KMeans

inertia = []

# With only 6 countries, test K from 2 to 5
K_range = range(2, 6)

for k in K_range:

    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    kmeans.fit(X_scaled)

    inertia.append(kmeans.inertia_)

plt.figure(figsize=(8, 5))

plt.plot(
    list(K_range),
    inertia,
    marker="o"
)

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method for K-Means")

plt.xticks(list(K_range))
plt.grid(True)

plt.show()
```

```python
from sklearn.metrics import silhouette_score

silhouette_scores = []

for k in range(2, 6):

    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = kmeans.fit_predict(X_scaled)

    score = silhouette_score(
        X_scaled,
        labels
    )

    silhouette_scores.append(score)

silhouette_results = pd.DataFrame({
    "K": range(2, 6),
    "Silhouette Score": silhouette_scores
})

silhouette_results
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    silhouette_results["K"],
    silhouette_results["Silhouette Score"],
    marker="o"
)

plt.xlabel("Number of Clusters")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Score for K-Means")

plt.xticks(range(2, 6))
plt.grid(True)

plt.show()
```

```python
#k-means

optimal_k = 3

kmeans = KMeans(
    n_clusters=optimal_k,
    random_state=42,
    n_init=10
)

cluster_labels = kmeans.fit_predict(X_scaled)

cluster_data["KMeans_Cluster"] = cluster_labels

cluster_data[
    ["location", "KMeans_Cluster"]
]
```

```python
#PCA visualisation

from sklearn.decomposition import PCA

pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_scaled)

pca_df = pd.DataFrame(
    X_pca,
    columns=["PC1", "PC2"]
)

pca_df["location"] = cluster_data["location"].values
pca_df["Cluster"] = cluster_labels

pca_df
```

```python
plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=pca_df,
    x="PC1",
    y="PC2",
    hue="Cluster",
    s=150
)

# Add country names
for i in range(len(pca_df)):
    plt.text(
        pca_df["PC1"].iloc[i],
        pca_df["PC2"].iloc[i],
        pca_df["location"].iloc[i],
        fontsize=9
    )

plt.title("K-Means Clustering of Countries")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")

plt.show()
```

```python
#hierarchial clustering

from scipy.cluster.hierarchy import linkage, dendrogram

linkage_matrix = linkage(
    X_scaled,
    method="ward"
)

plt.figure(figsize=(12, 6))

dendrogram(
    linkage_matrix,
    labels=cluster_data["location"].values,
    leaf_rotation=45
)

plt.title("Hierarchical Clustering Dendrogram")
plt.xlabel("Country")
plt.ylabel("Euclidean Distance")

plt.show()
```

```python
#Agglomerative Hierarchical Clustering

from sklearn.cluster import AgglomerativeClustering

hierarchical = AgglomerativeClustering(
    n_clusters=optimal_k,
    linkage="ward"
)

hierarchical_labels = hierarchical.fit_predict(
    X_scaled
)

cluster_data["Hierarchical_Cluster"] = hierarchical_labels

cluster_data[
    ["location", "KMeans_Cluster", "Hierarchical_Cluster"]
]
```
