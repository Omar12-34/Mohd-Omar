```python
#Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("ggplot")
```

```python
#Load Dataset
url = "https://raw.githubusercontent.com/salemprakash/EDA/main/Data/covid-data.csv"

covid = pd.read_csv(url)

covid.head()
```

```python
#Dataset Information
print("Shape:", covid.shape)

covid.info()

covid.describe(include="all")
```

```python
#Basic Statistical Analysis
covid.describe()

print("\nMean")
print(covid.mean(numeric_only=True))

print("\nMedian")
print(covid.median(numeric_only=True))

print("\nStandard Deviation")
print(covid.std(numeric_only=True))
```

```python
#Missing Values
print(covid.isnull().sum())

print("\nPercentage Missing")

print((covid.isnull().sum()/len(covid))*100)
```

```python
#Handle Missing Data
numeric_cols = covid.select_dtypes(include=np.number).columns

covid[numeric_cols] = covid[numeric_cols].fillna(
    covid[numeric_cols].median()
)

categorical_cols = covid.select_dtypes(include="object").columns

covid[categorical_cols] = covid[categorical_cols].fillna("Unknown")

covid.isnull().sum()
```

```python
#Data Cleaning
covid.drop_duplicates(inplace=True)

covid["date"] = pd.to_datetime(covid["date"])

covid = covid[
    (covid["total_cases"] >= 0) &
    (covid["total_deaths"] >= 0) &
    (covid["new_cases"] >= 0)
]

covid.head()
```

```python
#Data Transformation
covid["log_total_cases"] = np.log1p(covid["total_cases"])

covid["mortality_rate"] = (
    covid["total_deaths"] /
    covid["total_cases"]
) * 100

covid["mortality_rate"] = covid["mortality_rate"].fillna(0)

covid.head()
```

```python
#Univariate Analysis (Histogram)
plt.figure(figsize=(8,5))

sns.histplot(covid["total_cases"], bins=30)

plt.title("Distribution of Total Cases")

plt.show()
```

```python
#Univariate Analysis (Boxplot)
plt.figure(figsize=(8,5))

sns.boxplot(y=covid["total_deaths"])

plt.title("Boxplot of Total Deaths")

plt.show()
```

```python
#Univariate Analysis (Bar Chart)
plt.figure(figsize=(10,5))

covid["continent"].value_counts().plot(kind="bar")

plt.title("Countries by Continent")

plt.xticks(rotation=45)

plt.show()
```

```python
#Bivariate Analysis (Scatter Plot)
plt.figure(figsize=(8,6))

sns.scatterplot(
    data=covid,
    x="total_cases",
    y="total_deaths"
)

plt.title("Total Cases vs Total Deaths")

plt.show()
```

```python
#Bivariate Analysis (Boxplot)
plt.figure(figsize=(10,5))

sns.boxplot(
    data=covid,
    x="continent",
    y="new_cases"
)

plt.xticks(rotation=45)

plt.title("New Cases by Continent")

plt.show()
```

```python
#Bivariate Analysis (Bar Chart)
top10 = (
    covid.groupby("location")["total_cases"]
    .max()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(10,6))

top10.sort_values().plot(kind="barh")

plt.title("Top 10 Countries by Total Cases")

plt.xlabel("Total Cases")

plt.show()
```

```python
#Multivariate Analysis (Correlation Heatmap)
plt.figure(figsize=(10,8))

numeric = covid.select_dtypes(include=np.number)

sns.heatmap(
    numeric.corr(),
    cmap="coolwarm"
)

plt.title("Correlation Matrix")

plt.show()
```

```python
#Multivariate Analysis (Scatter Plot)
plt.figure(figsize=(10,6))

sns.scatterplot(
    data=covid,
    x="total_cases",
    y="total_deaths",
    hue="continent",
    alpha=0.6
)

plt.title("Cases vs Deaths by Continent")

plt.show()
```

```python
#Multivariate Analysis (Scatter Plot)
cols = [
    "total_cases",
    "total_deaths",
    "new_cases",
    "population"
]

sns.pairplot(covid[cols])

plt.show()
```
