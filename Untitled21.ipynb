```python
import os, json, random, re, zipfile
from pathlib import Path
from urllib.parse import urlsplit

import numpy as np
import pandas as pd
import tensorflow as tf
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    roc_auc_score, confusion_matrix, classification_report
)

SEED = 42
MAX_URL_LENGTH = 256
BATCH_SIZE = 64
EPOCHS = 12
OUTPUT_DIR = Path("artifacts")
OUTPUT_DIR.mkdir(exist_ok=True)

random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

TRAIN_FILE = "phreshphish_train.csv"
TEST_FILE = "phreshphish_test.csv"

def load_csv(path):
    df = pd.read_csv(path)
    if not {"url", "label"}.issubset(df.columns):
        raise ValueError(f"{path} must have columns 'url' and 'label'. Found: {list(df.columns)}")
    df = df[["url", "label"]].copy()
    df["url"] = df["url"].fillna("").astype(str).str.strip()
    df["label"] = pd.to_numeric(df["label"], errors="coerce")
    df = df[(df["url"] != "") & df["label"].isin([0, 1])].copy()
    df["label"] = df["label"].astype("int32")
    if df["label"].nunique() != 2:
        raise ValueError(f"{path} must contain both label 0 and label 1.")
    return df.reset_index(drop=True)

train_df = load_csv(TRAIN_FILE)
test_df = load_csv(TEST_FILE)
print("Train shape:", train_df.shape)
print("Test shape:", test_df.shape)
print("Training label counts (0=legitimate, 1=phishing):")
print(train_df["label"].value_counts().sort_index())
display(train_df.head())
```

```python
MAX_URL_LENGTH = 128
BATCH_SIZE = 32

def encode_urls(urls, max_length=MAX_URL_LENGTH):
    encoded = np.zeros((len(urls), max_length), dtype=np.uint16)

    for i, url in enumerate(urls):
        byte_values = str(url).encode("utf-8")[:max_length]
        if byte_values:
            encoded[i, :len(byte_values)] = (
                np.frombuffer(byte_values, dtype=np.uint8).astype(np.uint16) + 1
            )

    return encoded

X_train = encode_urls(train_part["url"].tolist())
X_val = encode_urls(val_part["url"].tolist())
X_test = encode_urls(test_df["url"].tolist())

y_train = train_part["label"].to_numpy(dtype=np.float32)
y_val = val_part["label"].to_numpy(dtype=np.float32)
y_test = test_df["label"].to_numpy(dtype=np.float32)

print("X_train:", X_train.shape)
print("X_val:", X_val.shape)
print("X_test:", X_test.shape)
```

```python
import re
from urllib.parse import urlparse
from sklearn.preprocessing import StandardScaler

def extract_features(url):
    url = str(url)
    lower = url.lower()

    try:
        parsed = urlparse(
            lower if "://" in lower else "//" + lower
        )
        hostname = parsed.hostname or ""
        path = parsed.path or ""
        query = parsed.query or ""
    except Exception:
        hostname, path, query = "", "", ""

    return [
        len(url),
        len(hostname),
        len(path),
        len(query),
        url.count("."),
        url.count("-"),
        url.count("_"),
        url.count("/"),
        url.count("@"),
        url.count("?"),
        url.count("="),
        url.count("&"),
        url.count("%"),
        sum(c.isdigit() for c in url),
        sum(c.isalpha() for c in url),
        sum(c.isupper() for c in url),
        int("https://" in lower),
        int("http://" in lower),
        int("@" in url),
        int("xn--" in lower),
        int(bool(re.fullmatch(r"\d{1,3}(?:\.\d{1,3}){3}", hostname))),
        int(any(word in hostname for word in [
            "login", "verify", "secure", "account", "update"
        ])),
    ]

feature_names = [
    "url_length", "hostname_length", "path_length", "query_length",
    "dots", "hyphens", "underscores", "slashes", "at_symbols",
    "question_marks", "equals", "ampersands", "percent_symbols",
    "digits", "letters", "uppercase", "has_https", "has_http",
    "has_at", "has_punycode", "has_ipv4_host", "suspicious_host_words"
]

F_train_raw = np.asarray(
    [extract_features(u) for u in train_part["url"]],
    dtype=np.float32
)
F_val_raw = np.asarray(
    [extract_features(u) for u in val_part["url"]],
    dtype=np.float32
)
F_test_raw = np.asarray(
    [extract_features(u) for u in test_df["url"]],
    dtype=np.float32
)

scaler = StandardScaler()
F_train = scaler.fit_transform(F_train_raw).astype(np.float32)
F_val = scaler.transform(F_val_raw).astype(np.float32)
F_test = scaler.transform(F_test_raw).astype(np.float32)

print("Feature count:", len(feature_names))
print("Feature matrix:", F_train.shape)
print("Features extracted successfully.")
```

```python
from tensorflow import keras
from tensorflow.keras import layers

def build_fast_model():
    url_input = keras.Input(
        shape=(MAX_URL_LENGTH,), dtype="int32", name="url_bytes"
    )
    feature_input = keras.Input(
        shape=(F_train.shape[1],), dtype="float32", name="url_features"
    )

    x = layers.Embedding(257, 32)(url_input)

    c3 = layers.Conv1D(32, 3, padding="same", activation="gelu")(x)
    c5 = layers.Conv1D(32, 5, padding="same", activation="gelu")(x)

    x = layers.Concatenate()([c3, c5])
    x = layers.Dense(64)(x)

    # One lightweight Transformer block
    z = layers.LayerNormalization()(x)
    a = layers.MultiHeadAttention(
        num_heads=2,
        key_dim=16,
        dropout=0.1
    )(z, z)

    x = layers.Add()([x, a])

    z = layers.LayerNormalization()(x)
    z = layers.Dense(128, activation="gelu")(z)
    z = layers.Dense(64)(z)
    x = layers.Add()([x, z])

    avg = layers.GlobalAveragePooling1D()(x)
    mx = layers.GlobalMaxPooling1D()(x)

    combined = layers.Concatenate()([avg, mx, feature_input])
    combined = layers.Dense(64, activation="gelu")(combined)
    combined = layers.Dropout(0.2)(combined)
    output = layers.Dense(1, activation="sigmoid")(combined)

    model = keras.Model(
        [url_input, feature_input], output
    )

    model.compile(
        optimizer=keras.optimizers.Adam(learning_rate=1e-4),
        loss="binary_crossentropy",
        metrics=[
            "accuracy",
            keras.metrics.Precision(name="precision"),
            keras.metrics.Recall(name="recall"),
            keras.metrics.AUC(name="auc")
        ]
    )

    return model

model = build_fast_model()
model.summary()
```

```python
import os
import gc

os.makedirs("artifacts", exist_ok=True)

checkpoint_path = "artifacts/best_hybrid_model.keras"

callbacks = [
    keras.callbacks.ModelCheckpoint(
        checkpoint_path,
        monitor="val_auc",
        mode="max",
        save_best_only=True,
        verbose=1
    ),
    keras.callbacks.EarlyStopping(
        monitor="val_auc",
        mode="max",
        patience=3,
        restore_best_weights=True,
        verbose=1
    ),
    keras.callbacks.ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.5,
        patience=2,
        min_lr=1e-6,
        verbose=1
    )
]

history = model.fit(
    [X_train.astype(np.int32), F_train],
    y_train,
    validation_data=(
        [X_val.astype(np.int32), F_val],
        y_val
    ),
    epochs=12,
    batch_size=32,
    callbacks=callbacks,
    verbose=1
)

gc.collect()
```

```python
import numpy as np
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)
from tensorflow import keras

checkpoint_path = "artifacts/best_hybrid_model.keras"

# Load the best checkpoint saved during training
model = keras.models.load_model(checkpoint_path)

# Predict phishing probabilities
test_probabilities = model.predict(
    [X_test.astype(np.int32), F_test],
    batch_size=32,
    verbose=1
).ravel()

# Convert probabilities into labels
test_predictions = (test_probabilities >= 0.5).astype(int)

print("TEST RESULTS")
print("=" * 45)
print(f"Accuracy:  {accuracy_score(y_test, test_predictions):.4f}")
print(f"Precision: {precision_score(y_test, test_predictions, zero_division=0):.4f}")
print(f"Recall:    {recall_score(y_test, test_predictions, zero_division=0):.4f}")
print(f"F1-score:  {f1_score(y_test, test_predictions, zero_division=0):.4f}")
print(f"ROC-AUC:   {roc_auc_score(y_test, test_probabilities):.4f}")

print("\nClassification Report:")
print(classification_report(
    y_test,
    test_predictions,
    target_names=["Legitimate", "Phishing"],
    digits=4,
    zero_division=0
))

print("Confusion Matrix:")
print(confusion_matrix(y_test, test_predictions))
```
